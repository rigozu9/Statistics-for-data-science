```python
alpha = 0.05
# True positive + False negative sum to 1, (AoS 10)
# False positive + True negative sum to 1, (AoS 10)

#Ex 1.1
power = 0.8
h0_holds = 100
h1_holds = 100
tp = h1_holds * power
fp = h0_holds * alpha
tn = h0_holds * (1 - alpha)
fn = h1_holds * (1 - power)
print(tp,fp,tn,fn)
```
