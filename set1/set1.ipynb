```python
mean = np.array([0, 1])
cov_matrix = np.array([[1, 0.5], [0.5, 2]])
x = stats.multivariate_normal(mean=mean, cov=cov_matrix).rvs(size=1000)
x1 = x[:,0]
x2 = x[:,1]
print(x1.shape, x2.shape)
x_summed = x1 + x2
x_summed_mean = x_summed.mean()
x_summed_var = x_summed.var()
print("Summed Mean of first 2 dimensions of X:", x_summed_mean)
print("Summed Variance of first 2 dimensions of X:", x_summed_var)
```

```python
## Ex 4.2
x1_mean = x1.mean()
x2_mean = x2.mean()
print("Mean of first dimension of X:", x1_mean)
print("Mean of second dimension of X:", x2_mean)
x1_x2_means_summed = x1_mean + x2_mean
print("Summed Mean of first 2 dimensions of X:", x1_x2_means_summed)
```

```python
x1_var = x1.var()
x2_var = x2.var()
print("Variance of first dimension of X:", x1_var)
print("Variance of second dimension of X:", x2_var)
x1_x2_vars_summed = x1_var + x2_var
print("Summed Variance of first 2 dimensions of X:", x1_x2_vars_summed)
```
