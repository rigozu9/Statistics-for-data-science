Ex 3
One-sided Wald test to assess whether its incidence of nausea is lower than the placebo's

```python
import math
from scipy.stats import norm
theta0 = 0

placebo_patients = 80
placebo_incidents = 45
p_placebo_hat = placebo_incidents / placebo_patients
placebo_side_of_se_hat = p_placebo_hat * (1 - p_placebo_hat) / placebo_patients

chlorpromazine_patients = 75
chlorpromazine_incidents = 26
p_chlorpromazine_hat = chlorpromazine_incidents / chlorpromazine_patients
chlorpromazine_side_of_se_hat = p_chlorpromazine_hat * (1 - p_chlorpromazine_hat) / chlorpromazine_patients

dimenhydrinate_patients = 85
dimenhydrinate_incidents = 52
p_dimenhydrinate_hat = dimenhydrinate_incidents / dimenhydrinate_patients
dimenhydrinate_side_of_se_hat = p_dimenhydrinate_hat * (1 - p_dimenhydrinate_hat) / dimenhydrinate_patients

pentobarbital_100_patients = 67
pentobarbital_100_incidents = 35
p_pentobarbital_100_hat = pentobarbital_100_incidents / pentobarbital_100_patients
pentobarbital_100_side_of_se_hat = p_pentobarbital_100_hat * (1 - p_pentobarbital_100_hat) / pentobarbital_100_patients

pentobarbital_150_patients = 84
pentobarbital_150_incidents = 36
p_pentobarbital_150_hat = pentobarbital_150_incidents / pentobarbital_150_patients
pentobarbital_150_side_of_se_hat = p_pentobarbital_150_hat * (1 - p_pentobarbital_150_hat) / pentobarbital_150_patients
```

```python
# Ex 3.1
# theta = p_drug - p_placebo
#H1 : theta < 0 means its incidence of nausea is lower than the placebo
# H0: theta >= 0 means drug has equal or higher nausea incidence than placebo (Lecture notes 7.3.1)

theta_hat_chlorpromazine = p_chlorpromazine_hat - p_placebo_hat

#Lecture notes 7.4.3 se_hat
se_hat_chlorpromazine = math.sqrt(chlorpromazine_side_of_se_hat + placebo_side_of_se_hat)

w = (theta_hat_chlorpromazine - theta0) / se_hat_chlorpromazine
print(w) # -2.7643637780027186

p_chlorpromazine = norm.cdf(w) # Lecture notes 7.4.1
print(p_chlorpromazine) #0.002852

# p < alpha, when alpha = 0.05, so we reject H0 and conclude that chlorpromazine has a lower incidence of nausea than the placebo
```

```python
# Ex 3.2
theta_hat_dimenhydrinate = p_dimenhydrinate_hat - p_placebo_hat

#Lecture notes 7.4.3 se_hat
se_hat_dimenhydrinate = math.sqrt(dimenhydrinate_side_of_se_hat + placebo_side_of_se_hat)

w = (theta_hat_dimenhydrinate - theta0) / se_hat_dimenhydrinate
print(w) # 0.6429872617820936

p_dimenhydrinate = norm.cdf(w) # Lecture notes 7.4.1
print(p_dimenhydrinate) #0.7398838172885269

# p >= alpha, when alpha = 0.05, There is not enough evidence to conclude that dimenhydrinate 100 mg has a lower incidence of nausea than placebo.
```

```python
# Ex 3.3
theta_hat_pentobarbital_100 = p_pentobarbital_100_hat - p_placebo_hat

#Lecture notes 7.4.3 se_hat
se_hat_pentobarbital_100 = math.sqrt(pentobarbital_100_side_of_se_hat + placebo_side_of_se_hat)

w = (theta_hat_pentobarbital_100 - theta0) / se_hat_pentobarbital_100
print(w) # -0.48642750289502995

p_pentobarbital_100 = norm.cdf(w) # Lecture notes 7.4.1
print(p_pentobarbital_100) #0.3133320473656943

# p >= alpha, when alpha = 0.05, There is not enough evidence to conclude that pentobarbital 100 mg has a lower incidence of nausea than placebo.
```

```python
# Ex 3.4
theta_hat_pentobarbital_150 = p_pentobarbital_150_hat - p_placebo_hat

#Lecture notes 7.4.3 se_hat
se_hat_pentobarbital_150 = math.sqrt(pentobarbital_150_side_of_se_hat + placebo_side_of_se_hat)

w = (theta_hat_pentobarbital_150 - theta0) / se_hat_pentobarbital_150
print(w) # -1.730218569934382


p_pentobarbital_150 = norm.cdf(w) # Lecture notes 7.4.1
print(p_pentobarbital_150) #0.041795615879166664

# p < alpha, when alpha = 0.05, so we reject H0 and conclude that pentobarbital_150 has a lower incidence of nausea than the placebo
```

Ex 4. 
Bonferroni and the FDR methods on the same dataset as in ex 3.

```python
alpha = 0.05
m = 4 # testsed 4 different drugs

# Ex 4.1.1 Bonferroni method of Chlorpromazine.
p_chlorpromazine_bonferroni = min(1, m * p_chlorpromazine)
print(p_chlorpromazine_bonferroni)
if p_chlorpromazine_bonferroni <= alpha: # (AoS 10.7)
    print("Reject H0")
else:
    print("Do not reject H0")
```

```python
# Ex 4.1.2 Bonferroni method of Dimenhydrinate.
p_dimenhydrinate_bonferroni = min(1, m * p_dimenhydrinate)
print(p_dimenhydrinate_bonferroni)
if p_dimenhydrinate_bonferroni <= alpha: # (AoS 10.7)
    print("Reject H0")
else:
    print("Do not reject H0")
```

```python
# Ex 4.1.3 Bonferroni method of Pentobarbital (100 mg).
p_pentobarbital_100_bonferroni = min(1, m * p_pentobarbital_100)
print(p_pentobarbital_100_bonferroni)
if p_pentobarbital_100_bonferroni <= alpha: # (AoS 10.7)
    print("Reject H0")
else:
    print("Do not reject H0")
```

```python
# Ex 4.1.4 Bonferroni method of Pentobarbital (150 mg).
p_pentobarbital_150_bonferroni = min(1, m * p_pentobarbital_150)
print(p_pentobarbital_150_bonferroni)
if p_pentobarbital_150_bonferroni <= alpha: # (AoS 10.7)
    print("Reject H0")
else:
    print("Do not reject H0")
```

```python
# Ex 4.2
# FDR method (AoS 10.7)
m = 4 # testsed 4 different drugs
p_val_dict = {
    "Chlorpromazine": p_chlorpromazine,
    "Dimenhydrinate": p_dimenhydrinate,
    "Pentobarbital 100 mg": p_pentobarbital_100,
    "Pentobarbital 150 mg": p_pentobarbital_150
}

sorted_p_val_dict = dict(sorted(p_val_dict.items(), key=lambda item: item[1]))

#print(sorted_p_val_dict)

adjusted_scaled_p_vals = {}

for j, (drug, p) in enumerate(sorted_p_val_dict.items(), 1):
    scaled = min(1, (m / j) * p)
    adjusted_scaled_p_vals[drug] = scaled

print("BH scaled p-values:", adjusted_scaled_p_vals)

```

Ex 5
Compare Bayesian and maximum likelihood estimation for the multinomial distribution.

```python
n = 20 # twenty throws
throws = [6, 3, 2, 1, 6, 6, 2, 6, 1, 3, 3, 2, 2, 2, 5, 3, 1, 6, 5, 1]
#print(len(throws))

count_dict = {str(i): 0 for i in range(1, 7)}

for throw in throws:
    count_dict[str(throw)] += 1
#print(count_dict)

count = 0
p_count_dict = count_dict.copy()
for key, value in p_count_dict.items():
    p_count_dict[key] = value / n
    count += value / n

print(p_count_dict,"this should be 1: ",count)

```

```python
# Ex 5.2
# Compute the predictive likelihood (i.e. likelihood of future observations) of observing the following sequence in the future: (1,5,4)
# From 6.7 Connection to Machine Learning: Cross-Entropy as MLE (lecture notes)
# Likelihood for categorical data: L_n(theta) = PI n, i = 1 Categorical(Y_i;f(Xi;theta))
likelihood = count_dict['1'] * count_dict['5'] * count_dict['4']
print(likelihood) # 0, 4 didnt appear once so it p_hat = 0, which makes it 0 in dot rule for likelihood.
```

```python
# Ex 5.3
# p ~ Dirichlet(alpha), E(p_i)=alpha_i/alpha0, where alpha0 = SIGMA k, j=1 alpha_j
# marginal distribution of p_i is Beta(alpha_i, alpha_0-alpha_i)
# X_1,...,X_n, the posterior of p is p| X_1,...,X_n ~Dirichlet(alpha*), with alpha* = (alpha_1*,...,alpha_k*), a_j*=a_j + SIGMA n, i=1 I(X_i=j)
```

```python
#i. alpha_1 = ... = alpha_k = 0.1
alpha = 0.1
zero1_count_dict = count_dict.copy()
for key, value in zero1_count_dict.items():
    zero1_count_dict[key] = value + alpha

#print(zero1_count_dict)

alpha0_star = sum(zero1_count_dict.values())
print(alpha0_star) # 20.6

posterior = {}
count = 0
for key, value in zero1_count_dict.items():
    posterior[key] = value / alpha0_star
    count += value / alpha0_star

print(posterior, count)
```

```python
#ii. alpha_1 = ... = alpha_k = 1
alpha = 1
zero1_count_dict = count_dict.copy()
for key, value in zero1_count_dict.items():
    zero1_count_dict[key] = value + alpha

#print(zero1_count_dict)

alpha0_star = sum(zero1_count_dict.values())
print(alpha0_star) #26

posterior = {}
count = 0
for key, value in zero1_count_dict.items():
    posterior[key] = value / alpha0_star
    count += value / alpha0_star

print(posterior, count)
```

```python
#iii. alpha_1 = ... = alpha_k = 10
alpha = 10
zero1_count_dict = count_dict.copy()
for key, value in zero1_count_dict.items():
    zero1_count_dict[key] = value + alpha

#print(zero1_count_dict)

alpha0_star = sum(zero1_count_dict.values())
print(alpha0_star) #80

posterior = {}
count = 0
for key, value in zero1_count_dict.items():
    posterior[key] = value / alpha0_star
    count += value / alpha0_star

print(posterior, count)
```

With smaller alphas the posterior stays way more close to what was found in 5.1 and larger alphas go toward equal probalities 1/6. I feel like using alpha = 1 prior for if we dont know if a die is fair or loaded is the best bet. It feels like the most balanced one.
